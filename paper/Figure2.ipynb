```python
starspot = (0.4,15)
import sys
import os
import pickle
methods = [
    "commensurate_x_virt_z_new",
    "magnus1_x_virt_z_nooptim",
    "FAST-MAGNUS-MAGNUS-fix1707-skipFM-N5",
    "FAST-MAGNUS-MAGNUS-fix1707-woFAST-amp-N5",
]
sys.path.append(os.path.abspath(".."))
gates = []
loadsuccess = False
if os.path.exists("gates.pickle"):
    try:
        with open("gates.pickle", "rb") as f:
            gates = pickle.load(f)
        loadsuccess = True
    except Exception as e:
        print(f"Error loading gates: {e}")
        loadsuccess = False
if not loadsuccess:
    print("Generating gates...")
    for method in methods:
        from bachelor.utils import eval_functions as ef
        from bachelor.utils import gate_params as gp
        import numpy as np
        learner = ef.AdaptiveLearner([],[None, 1, None, 1, (starspot[0]*0.9999,starspot[0]*1.00001), np.pi, (starspot[1]*0.9999,starspot[1]*1.00001), np.inf, np.inf],framework="random",truncation=5)
        qb = learner.get_next_dp(N=1)[0]
        gate = gp.get_gate_params(method)
        qb,gate = ef.calib_gate((qb,gate))
        gates.append(gate)
    with open("gates.pickle", "wb") as f:
        pickle.dump(gates, f)
```
