```python
import os
import numpy as np
import pickle

folder = os.listdir()
data4fig = ""
for file in folder:
    if file.startswith("Fig3data") and file.endswith(".pickle"):
        if "_temp" in file and data4fig == "":
            data4fig = file
        elif "_temp" not in file:
            data4fig = file
with open(data4fig, "rb") as f:
    data4fig = pickle.load(f)

targetDataset = "FAST-MAGNUS-MAGNUS-fix1707-amp-woFAST-derrquad-N3"
data4fig = data4fig[targetDataset]

for i,point in enumerate(data4fig):
    newpoint = {}
    newpoint["alpha"] = point["alpha_actual"]/point["omega_01_actual"]
    newpoint["Lambdas"] = point["Lambdas"]
    newpoint["score"] = point["score"]
    data4fig[i] = newpoint
    

print("Data loaded succesfully.")
```
