```python
with open("tex.txt", "r") as f:
    tex = f.read()
tex_tmp = tex.split("%")
tex = tex_tmp[0]
for line in tex_tmp[1:]:
    line = line.split("\n")
    for l in line:
        tex += l
tex = tex.replace("\n", "")
tex = tex.replace("\\right.$$\item$$\left.", "")
tex = tex.replace("\left(", "(")
tex = tex.replace("\\right)", ")")
tex = tex.replace("\left[", "(")
tex = tex.replace("\\right]", ")")
tex = tex.replace("\item$$", "")
tex = tex.replace("$$", "")
tex = tex.replace("\\right.", "")
tex = tex.replace("\\begin{enumerate}", "")
tex = tex.replace("\end{enumerate}", "")
tex = tex.replace("\sum_{n,m=1}", "")
print(tex)
```
